```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time

print("TensorFlow 버전:", tf.__version__)
print("GPU 사용 가능:", tf.config.list_physical_devices('GPU'))
```

```python
# 데이터 로드 (CIFAR-10)

(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

y_train = y_train.flatten()
y_test = y_test.flatten()

CLASS_NAMES = ["airplane", "automobile", "bird", "cat", "deer",
               "dog", "frog", "horse", "ship", "truck"]

print(f"학습 데이터: {x_train.shape}, 테스트 데이터: {x_test.shape}")
```

```python
# 모델 정의 - Sequential API 버전

def build_model(num_classes=10):
    model = keras.Sequential([
        layers.Input(shape=(32, 32, 3)),

        layers.Conv2D(32, 3, padding="same", activation="relu"),
        layers.MaxPooling2D(2),

        layers.Conv2D(64, 3, padding="same", activation="relu"),
        layers.MaxPooling2D(2),

        layers.Conv2D(64, 3, padding="same", activation="relu"),

        layers.Flatten(),
        layers.Dense(64, activation="relu"),
        layers.Dense(num_classes, activation="softmax"),
    ])
    return model


model = build_model()
model.summary()  # README 표에 넣을 파라미터 수 확인용
```

```python
# 컴파일 & 학습

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

start = time.time()

history = model.fit(
    x_train, y_train,
    validation_data=(x_test, y_test),
    epochs=10,
    batch_size=64,
)

elapsed = time.time() - start
print(f"\n총 학습 시간: {elapsed:.1f}초")

```


fit() 대신 직접 학습 루프를 짜고 싶을 때 Custom Training Loop 버전 - GradientTape 을 사용할 수 있음.

```python
model.save("cnn_keras.keras")
print("모델 저장 완료: cnn_keras.keras")
```
